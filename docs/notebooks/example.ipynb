# A Wiz notebook

This notebook is deliberately small.  It has a definition, two ways of printing a
value, an error, and a `%%scry` cell.  Running `wiz nb export` turns the stored
outputs below into one static HTML page: no kernel, no JavaScript, no network.

A definition stays in scope for the cells after it.  With the `repl` driver that is
a live process; with the `batch` fallback the notebook is rechecked from the top,
so the meaning is the same even though the state is not.


```wiz
def double (n : Nat) : Nat := n + n
```

`#eval` is a command: its value is a stream, exactly as in a file.


```wiz
#eval double 21
```

A bare last expression becomes the cell's *result*.  This is the kernel's
convenience, not new Wiz semantics: a Wiz file has no expression statements.


```wiz
20 + 22
```

A Wiz error is reported on the error channel, and the session goes on.


```wiz
def broken : Nat := Bool.true
```

`%%scry` reads the cell as C source and renders the taint findings.  It needs
a `scry` executable (`PATH`, or `$WIZ_SCRY`); without one the cell reports
`ScryUnavailable` instead of failing the kernel.


```wiz
%%scry
char *getenv(const char *name);
int system(const char *command);

int main(void) {
  char *cmd = getenv(0);
  system(cmd);
  return 0;
}
```
